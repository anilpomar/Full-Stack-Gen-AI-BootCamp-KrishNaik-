```python
#Helper Function -This Function Prints the chunks consistently and collect simple statistics

from typing import Any

def print_chunks(
    title: str,
    chunks: list[Any],
    preview_limit: int = 500,
) -> None:
    print(f"\n{'=' * 90}")
    print(title)
    print(f"Total chunks: {len(chunks)}")
    print("=" * 90)

    for index, chunk in enumerate(chunks, start=1):

        # Get text from a LangChain Document or normal Python object
        text = getattr(
            chunk,
            "page_content",
            str(chunk),
        )

        # Get metadata if available
        metadata = getattr(
            chunk,
            "metadata",
            {},
        )

        print(
            f"\nChunk {index} | "
            f"characters={len(text)}"
        )

        if metadata:
            print("Metadata:", metadata)

        if len(text) > preview_limit:
            print(text[:preview_limit] + "...")
        else:
            print(text)
```

```python
from typing import Any
import numpy as np


def chunk_stats(
    method: str,
    chunks: list[Any],
) -> dict[str, Any]:

    lengths = [
        len(
            getattr(
                chunk,
                "page_content",
                str(chunk),
            )
        )
        for chunk in chunks
    ]

    return {
        "method": method,
        "number_of_chunks": len(chunks),
        "minimum_length": min(lengths) if lengths else 0,
        "maximum_length": max(lengths) if lengths else 0,
        "average_length": (
            round(float(np.mean(lengths)), 2)
            if lengths
            else 0
        ),
        "total_stored_characters": sum(lengths),
    }
```

## Fixed-Size Chunking

1. This Chunking TEchnique divides content after a specified number of characters or tokens
2. It is simple and Predictable And it may Break Words,Sentences,Paragraphs, Logical Topics

```python
def fixed_size_chunking(text: str, chunk_size: int, chunk_overlap: int = 0) -> list[str]:
    if chunk_size <= 0:
        raise ValueError("chunk_size must be greater than zero.")

    if chunk_overlap < 0 or chunk_overlap >= chunk_size:
        raise ValueError(
            "chunk_overlap must satisfy 0 <= chunk_overlap < chunk_size."
        )

    step = chunk_size - chunk_overlap

    return [
        text[start:start + chunk_size]
        for start in range(0, len(text), step)
    ]
```

## RecursiveCharacterTextSplitter

1. This Splitter Tries seperators Recursively.
2. Paragraph → line → sentence-like boundary → word → character.

What is RecursiveCharacterTextSplitter?
- Recursive chunking decides where to split.
- LangChain's RecursiveCharacterTextSplitter works by trying a hierarchy of separators.

***Typically***

Full Document
     │
     ▼
Try "\n\n" → split by paragraphs
     │
     ▼
If chunks are still too large
Try "\n" → split by lines
     │
     ▼
Still too large?
Try " " → split by words
     │
     ▼
Still too large?
Split by characters           

## *How Internally RecursiveCharacterText Splitter Works

Large Text
   │
   ▼
Try paragraph separator \n\n
   │
   ├── Small enough → keep
   │
   └── Too large
          │
          ▼
      Try \n
          │
          ├── Small enough → keep
          │
          └── Too large
                 │
                 ▼
               Try " "
                 │
                 ▼
              Split again

***Suppose:***
RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20
)

***Steps of Execution***
1.  It first tries to split using:   \n\n    
   Then: If the resulting pieces are still larger than 100 characters, it recursively tries using: \n
   Then: If still the chunksize doesn't satisfy it will recursively tries using :" "
   Then: And eventually individual characters if necessary.

***Why it is commonly used***
1. It works well for general text
2. The goal is to preserve natural text structure as much as possible.
3. It still controls chunk size
4. It is a strong baseline for PDFs, reports, policies, and articles

***Rule***
Recursive "Preserve document structure as much as possible

```python
recursive_splitter = RecursiveCharacterTextSplitter(
    separators=["\n\n", "\n", ". ", " ", ""],
    chunk_size=300,
    chunk_overlap=50,
    length_function=len,
    add_start_index=True,
)
```

## TokenTextSplitter

LLMs and Embeddings Models process Token rather than raw characters.


What is TokenTextSplitter ?
1. Instead of Chunking Text here splitter creates Chunks on Tokens.
2. Here Chunks are created on Tokens not on Passed Text.
    - So chunk_size=100 means approximately 100 tokens per chunk, not 100 characters or 100 words.

How TokenTextSplitter Works
- Text → Tokenization → Split tokens into chunks


- Visualization
Original Text
      ↓
Tokenizer
      ↓
Tokens
      ↓
[Token 1 ... Token 100]       ← Chunk 1
          ↕ 20 token overlap
[Token 81 ... Token 180]      ← Chunk 2
          ↕ 20 token overlap
[Token 161 ... Token 260]     ← Chunk 3


**Token Bases Token used when?**
1. When the Embeddings Model has a strict Token .
2. You need predictable model-input sizing.
3. Character counts do not accurately represent model usuage. (Character count is not token count)
    - Take Text:internationalization it could represent 19characters but necessarily not 19Tokens it may get 50 tokens. So like this if u have 4-5 words it will be eventually breach Embeddings model (Token) Size.

**Token count varies with:**
- Language
- Punctuation
- Numbers
- Source code
- Model tokenizer

***Note: Recursive Chunking can also use Tokens.***
- RecursiveCharacterTextSplitting can be configured with Token-Based Length function.
- Benefit  : 
Recursive splitting strategy
        +
Token-based measurement

***Rule***
Token-"I care about the number of tokens sent to the LLM."

```python
#In This example if have used cl100k_base tokenizer.
token_splitter = TokenTextSplitter(
    encoding_name="cl100k_base",
    chunk_size=80,
    chunk_overlap=15,
)
```
